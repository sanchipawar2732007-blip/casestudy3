```python
# Stack Implementation using Linked List

class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


class Stack:
    def __init__(self):
        self.top = None

    # Push operation
    def push(self, data):
        new_node = Node(data)
        new_node.next = self.top
        self.top = new_node
        print(data, "pushed into stack")

    # Pop operation
    def pop(self):
        if self.top is None:
            print("Stack is empty!")
        else:
            print("Popped element:", self.top.data)
            self.top = self.top.next

    # Peek operation
    def peek(self):
        if self.top is None:
            print("Stack is empty!")
        else:
            print("Top element:", self.top.data)

    # Display stack
    def display(self):
        if self.top is None:
            print("Stack is empty!")
            return

        temp = self.top
        print("Stack elements (Top to Bottom):", end=" ")

        while temp is not None:
            print(temp.data, end=" ")
            temp = temp.next

        print()


# Main program
s = Stack()

s.push(10)
s.push(20)
s.push(30)

s.display()
s.peek()
s.pop()
s.display()
```
